```python
def simple_chatbot(user_query):
    # Padroniza a entrada do utilizador para minúsculas
    query = user_query.strip().lower()
    
    if "total revenue" in query or "receita" in query:
        return "For Microsoft in 2025, the total revenue reached $281,200 million (a growth of ~14.7% compared to 2024)."
    elif "net income" in query or "lucro" in query:
        return "For Microsoft in 2025, net income was $101,800 million (an increase of ~15.5% from the previous year)."
    elif "cash flow" in query or "fluxo de caixa" in query:
        return "For Microsoft in 2025, operating cash flow reached $115,000 million."
    elif "assets" in query or "ativos" in query:
        return "For Microsoft in 2025, total assets were $510,000 million against $255,000 million in total liabilities."
    else:
        return "Sorry, I can only provide information on predefined queries (total revenue, net income, cash flow, assets)."

if __name__ == "__main__":
    print("--- BCG Financial AI Chatbot (Prototype) ---")
    print("Type your question or type 'exit' to quit.\n")
    
    # Exemplo de teste simulado ou loop de conversação
    test_queries = [
        "What is the total revenue?",
        "How has net income changed?",
        "What about cash flow?",
        "Unknown question test"
    ]
    
    for q in test_queries:
        print(f"User: {q}")
        print(f"Chatbot: {simple_chatbot(q)}\n")
```
