```python
import pandas as pd

data = {
    'Company': ['Microsoft', 'Microsoft', 'Microsoft', 'Tesla', 'Tesla', 'Tesla', 'Apple', 'Apple', 'Apple'],
    'Year': [2023, 2024, 2025, 2023, 2024, 2025, 2023, 2024, 2025],
    'Total Revenue': [211900, 245100, 281200, 96800, 97700, 102500, 383300, 391000, 410500],
    'Net Income': [72400, 88100, 101800, 15000, 7100, 9500, 97000, 93700, 100200],
    'Total Assets': [364800, 484300, 510000, 106600, 122200, 135000, 352600, 352500, 360000],
    'Total Liabilities': [205800, 243000, 255000, 43000, 47900, 52000, 290400, 294200, 298000],
    'Operating Cash Flow': [87600, 102600, 115000, 13300, 13800, 15200, 110500, 118200, 125000]
}

df = pd.DataFrame(data)

df['Revenue Growth (%)'] = df.groupby('Company')['Total Revenue'].pct_change() * 100
df['Net Income Growth (%)'] = df.groupby('Company')['Net Income'].pct_change() * 100

print("--- ANÁLISE DE TENDÊNCIAS FINANCEIRAS ---")
display(df[['Company', 'Year', 'Total Revenue', 'Revenue Growth (%)', 'Net Income', 'Net Income Growth (%)']])
```
